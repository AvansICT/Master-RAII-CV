## 🎥 Stap 2: Webcam detectie

In deze code gebeurt het volgende:

1. We openen de webcam van je computer  
2. We lezen continu beelden (frames) in een loop  
3. Voor elk frame:
   - voert het YOLO-model objectdetectie uit  
   - tekent het model bounding boxes om objecten  
4. We tonen het resultaat in een venster  
5. Als je op **'q'** drukt, stopt het programma  

💡 Dit heet **real-time inference**: het model maakt voorspellingen terwijl de video loopt.

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)
    annotated = results[0].plot()

    cv2.imshow("YOLO Webcam", annotated)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
